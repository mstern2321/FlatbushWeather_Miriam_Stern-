Question

Write:

def find_shared_students(students):

The parameter students is a list of Student objects.
Return a set of student names who share at least one course with another student.

Requirements:

Use nested loops
Use shares_course_with()
Avoid duplicate comparisons (comparing more than once with the same object)
Return a set


Example Usage / Expected Output
s1 = Student("Avi")
s2 = Student("Leah")
s3 = Student("Noah")
s4 = Student("Sara")

s1.add_course("Math")
s1.add_course("CS")
s2.add_course("Math")
s3.add_course("CS")
s3.add_course("History")
s4.add_course("Art")

students = [s1, s2, s3, s4]

print(find_shared_students(students))
Expected output:

{'Avi', 'Leah', 'Noah'}   # order may vary because this is a set
Explanation: Avi shares Math with Leah and CS with Noah. Sara shares no courses.

```python

def find_shared_students(students):
    my_set = set()

    for i in range(len(students)-1):
      for student in students:
        other_student = students[i+1]
        if student.shares_course_with(other_student) and student._name != (other_student._name):
          my_set.add(student._name)
          my_set.add(other_student._name)

    return my_set

s1 = Student("Avi")
s2 = Student("Leah")
s3 = Student("Noah")
s4 = Student("Sara")

s1.add_course("Math")
s1.add_course("CS")
s2.add_course("Math")
s3.add_course("CS")
s3.add_course("History")
s4.add_course("Art")

students = [s1, s2, s3, s4]

print(find_shared_students(students))
```

```python
def build_course_roster(students):
  my_dict = {}
  for student in students:
    for course in student._courses:
      if course not in my_dict:
        my_dict[course] = set()
      my_dict[course].add(student._name)
  return my_dict
s1 = Student("Avi")
s2 = Student("Leah")
s3 = Student("Noah")

s1.add_course("Math")
s1.add_course("CS")
s2.add_course("Math")
s2.add_course("Bio")
s3.add_course("CS")
s3.add_course("History")

students = [s1, s2, s3]

print(build_course_roster(students))
```
